# Similarity and Clustering (Unsupervised Learning)
> 核心概念：用相似度来度量文本之间的相似性，进而进行聚类
> Use similarity-based features to measure the similarity between texts, and then cluster them.
### 1. preprocess text with `clean_text`
### 2. ITF-IDF vectorization: convert the text into a vector representation把目标文本换成数值矩阵
### 3. Document similarity: calculate the similarity between documents 文档相似度
- 常用方法：cosine similarity `sklearn.metrics.pairwise.cosine_similarity`
    > Treat each document as a vector, and we calcualte their cosine，夹角越小越相似，数值越大越相似
- 其他方法：Euclidean, Manhattan, BM25, Jaccard \
### 4. Clustering: group similar documents together
- Clustering is an unsupervised learning that groups similar documents together, 利用相似度矩阵进行聚类
- 常用不同分法：**Agglomerative Clustering**, **Divisive: top-down** \
    - Agglomerative: bottom-up（自底向上），先把每个文档看成一个 cluster，然后合并相近的 \
    - Divisive: top-down（自顶向下），从全集开始，把最不相似的文档不断拆分 \
### 5. Linkage: 连接方式
- 连接方式决定了如何计算文档之间的距离 \
- 常用的连接方式有：`single linkage`, `complete linkage`, `average linkage`, `ward`
- input: technically it should be distance matrix, but the professor uses similarity matrix
    - **Ward**: Ward’s minimum variance method is a criterion to minimize total within-cluster variance. At each step, we find the pair of clusters that leads to the minimum increase in total within-cluster variance after merging.
    - **Single linkage**: Based on the nearest-neighbor, or minimum distance between records; tends to form long slender clusters
    - **Complete linkage**: Based on the farthest-neighbor, or maximum distance between records; tends to form compact sphere-like clusters
    - **Average linkage**: Based on the average distance between all records in one cluster to all records in another cluster; reduces dependence on extreme values (such as most similar or dissimilar records)
    > 它们都需要 簇之间的“距离”，不是“相似度”！\
    > Ward minimizes the total within-cluster distance squared (like variance). \
    > Complete linkage looks at the maximum distance between points in two clusters. \
    > Single linkage looks at the minimum distance between points in two clusters. \
    > Average linkage computes the average distance between all point-pairs across clusters.







# Topic Modeling
> Clustering是把数据按“相似性”分几类:
> > - 特征：
> > - 输入：数值向量（可以是图像、TF-IDF、word embeddings等）
> > - 输出：每个样本的簇标签（例如文档 A 属于cluster 2)
> 而Topic Modeling是把数据按主题分几类
> > - 特征：
> > - 输入：一堆原始文本，没有任何文档的“正确主题”标签
> > - 模型（如 LDA、LSI）会自己找出一些“隐藏主题”并告诉你：每篇文档属于哪些主题、每个主题有哪些关键词
| 比较点 | Topic Modeling | Clustering |
|--------|----------------|------------|
| 学习类型 | Unsupervised | Unsupervised |
| 输出是什么 | 文档 → 多个主题（概率分布） | 文档 → 只属于一个簇（类别） |
| 背后思路 | 找出“隐藏的语义变量”（主题） | 找出“相似的点簇”（密度/距离） |
| 示例模型 | LDA, LSI, NMF | KMeans, Hierarchical, DBSCAN |
| 一篇文档可以有多个标签吗？ | ✅ 是的（比如：30% 在 topic 1，70% 在 topic 2） | ❌ 一般只有一个cluster标签 |
